## invoice_id, track_id, track_name, media_type name, quantity, unit_price where invoice_id = 1

```python
q=""" select il.invoice_id , t.track_id , t.name as track_name, md.name as media_type_name , il.quantity , il.unit_price 
    from invoice_line il
    inner join track t
    on
    il.track_id=t.track_id
    inner join media_type md
    on
    t.media_type_id=md.media_type_id
    where il.invoice_id == 1
"""
sql_query(q)
```

## invoice_id, track_id, track_name, artist_name, media_type_name, quantity, unit_price, where invoice_id = 1

```python
q=""" select il.invoice_id , t.track_id , t.name as track_name,ar.name as artist_name, md.name as media_type_name , il.quantity , il.unit_price 
    from invoice_line il
    inner join track t
    on
    il.track_id=t.track_id
    inner join album al
    on
    t.album_id = al.album_id
    inner join artist ar
    on
    al.artist_id = ar.artist_id
    inner join media_type md
    on
    t.media_type_id=md.media_type_id
    where il.invoice_id == 1
"""
sql_query(q)
```

## track_id, album title, artist name

```python
q=""" select t.track_id ,al.title album_title , ar.name artist_name
    from track t 
    inner join album al
    on
    t.album_id = al.album_id
    inner join artist ar
    on
    al.artist_id = ar.artist_id
"""
sql_query(q)
```

## no_of_tracks_sale, album_name, artist_name

if we want to save extracted information in csv file so we can do it.

```python
q =""" select count(il.track_id) No_of_tracks_sale,al.title album_name,ar.name artist_name
        from invoice_line il
        inner join track t
        on
        il.track_id = t.track_id
        inner join album al
        on
        t.album_id=al.album_id
        inner join artist ar
        on
        al.artist_id=ar.artist_id
        group by 2
        order by 1 DESC limit 5
        
"""
df=sql_query(q)
df.to_csv("Top_5_sales.csv")
```
